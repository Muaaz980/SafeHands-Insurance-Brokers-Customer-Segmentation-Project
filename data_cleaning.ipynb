```python
import pandas as pd
```

```python
# Load your insurance dataset (ensure 'insurance.csv' is in your working directory)
df = pd.read_csv('insurance.csv')
```

```python
# Display the first 5 rows to verify columns
print("First 5 rows of the dataset:")
display(df.head())
```

```python
# Check shape and data types
print("\nDataset Info:")
df.info()
```

```python
# Count missing values per column
missing_values = df.isnull().sum()
print("Missing values per column:")
print(missing_values)
```

```python
# Summary statistics for numerical columns (check for weird ranges like negative age or BMI)
print("Numerical Summary:")
display(df.describe())
```

```python
# Check unique values in categorical columns to spot typos or malformed entries
print("\nUnique values in categorical columns:")
for col in ['sex', 'smoker', 'region']:
    print(f"{col}: {df[col].unique()}")
```

```python
# Drop rows where any column has missing values (safe fallback)
initial_row_count = len(df)
df = df.dropna()
dropped_rows = initial_row_count - len(df)

print(f"Dropped {dropped_rows} rows containing missing values.")
print(f"Current dataset shape: {df.shape}")
```

```python
# Convert all column names to lowercase, replacing spaces or special characters with underscores
df.columns = df.columns.str.lower().str.replace(' ', '_').str.replace('-', '_')

print("Updated column names:", df.columns.tolist())
```

```python
# Define categorical columns and convert them to 'category' type
categorical_cols = ['sex', 'smoker', 'region']
for col in categorical_cols:
    if col in df.columns:
        df[col] = df[col].astype('category')

# Ensure numerical columns have proper numeric types
if 'age' in df.columns:
    df['age'] = df['age'].astype(int)
if 'children' in df.columns:
    df['children'] = df['children'].astype(int)
if 'bmi' in df.columns:
    df['bmi'] = df['bmi'].astype(float)
if 'charges' in df.columns:
    df['charges'] = df['charges'].astype(float)

# Verify the updated data types
print("\nUpdated Data Types:")
print(df.dtypes)
```

```python
# Save the cleaned dataframe to a CSV file (without index column)
df.to_csv('cleaned_customers.csv', index=False)

print("Success! Cleaned dataset saved as 'cleaned_customers.csv'.")
```
